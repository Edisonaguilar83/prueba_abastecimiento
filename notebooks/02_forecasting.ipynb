# 02 — Forecasting y validación temporal

Este notebook ejecuta el mismo backtest que el pipeline: rolling-origin, sin métricas hard-codeadas.


```python
from pathlib import Path
import pandas as pd
from src.forecasting import backtest_forecasting

ROOT = Path.cwd()
if not (ROOT / 'data').exists(): ROOT = ROOT.parent
daily = pd.read_csv(ROOT/'data/processed/daily/dataset_daily_features.csv', parse_dates=['fecha'])
origin = daily.fecha.max()
origins = pd.date_range(origin - pd.Timedelta(days=42), origin - pd.Timedelta(days=7), freq='7D')
oof, metrics = backtest_forecasting(daily, origins)
metrics
```

```python
base = metrics.iloc[0]
metrics.assign(
    rmse_improvement_vs_baseline=(base.RMSE-metrics.RMSE)/base.RMSE,
    wape_improvement_vs_baseline=(base.WAPE-metrics.WAPE)/base.WAPE
)
```
